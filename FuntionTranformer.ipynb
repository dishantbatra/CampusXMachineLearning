```python
import numpy as np
import pandas as pd

df = pd.read_csv("/content/train.csv",usecols=['Age','Fare','Survived'])
df.head()
#

df.isnull().sum()
```

```python
#since age has some null values we can fill with the average value

df['Age'].fillna(df['Age'].mean(),inplace=True)
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score
from sklearn.model_selection import cross_val_score
import seaborn as sns
```

```python
df.describe()
```

```python
import matplotlib.pyplot as plt
import scipy.stats as stats

plt.figure(figsize=(14,4))
plt.subplot(231)
sns.kdeplot(df['Age'])

# print(df['Age'].skew())
plt.title('Age')

plt.subplot(232)
sns.kdeplot(df['Fare'])
# print(df['Fare'].skew())
plt.title('Fare')

plt.subplot(233)
sns.kdeplot(df['Survived'])
plt.title('Survived')
# print(df['Survived'].skew())

#stats.probplot (from scipy.stats) is used to create data for a Q–Q plot (quantile–quantile plot).
#A Q–Q plot helps you check whether your data follows a normal distribution (or another theoretical distribution).

plt.subplot(234)
stats.probplot(df['Age'],dist="norm",plot=plt)
plt.title('Age Prob')

plt.subplot(235)
stats.probplot(df['Fare'],dist="norm",plot=plt)
plt.title('Fare Prob')

plt.subplot(236)
stats.probplot(df['Survived'],dist="norm",plot=plt)
plt.title('Survived Prob')

plt.subplots_adjust(hspace=0.7, wspace=0.3)

plt.show()
```
